# geo4bolivia: find, load, map and analyse Bolivia’s subnational data

**Quickstart without the package (plain URLs)**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/tools/geo4bolivia-package/](https://quarcs-lab.github.io/gdo4bolivia/tools/geo4bolivia-package/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
import pandas as pd
import plotly.express as px

# The same files without the package: plain URLs
BASE = "https://raw.githubusercontent.com/quarcs-lab/gdo4bolivia/main/"
df = pd.read_csv(BASE + "data/sdg01/map343/sdg01.csv")
units = pd.read_csv(BASE + "geo/units.csv")      # names, parents, INE codes
units = units[units["scale"].isin(["map343", "department"])]   # sdg01's map
df = df.merge(units[["geo_id", "name", "department_id"]], on="geo_id")
df["department"] = df["department_id"].map(units.set_index("geo_id")["name"])
df["period"] = df["year"].astype(str)
px.box(df, x="department", y="sdg1_2_mpi", color="period",
       labels={"sdg1_2_mpi": "Multidimensional poverty index (0-1)"},
       title="Multidimensional poverty by department, ~2012 and ~2024")
```
