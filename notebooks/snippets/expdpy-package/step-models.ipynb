# expdpy: explore and model panel data

**Pooled, between, fixed and random effects**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/tools/expdpy-package/](https://quarcs-lab.github.io/gdo4bolivia/tools/expdpy-package/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
!pip install -q geo4bolivia expdpy
import numpy as np
import geo4bolivia as geo
import expdpy as ex

ids = geo.load("sdg", variables=["index_sdg", "population"])
ids["ln_pop"] = np.log(ids["population"])
df, df_def = geo.for_expdpy(ids, labels={"ln_pop": "Log of census population"})
df = ex.set_labels(df, df_def, set_panel=True)
print(ex.analyze_hausman_test(df, "index_sdg", ["ln_pop"]).interpret())
ex.analyze_panel_table(df, "index_sdg", ["ln_pop"]).etable
```
