# Population

**Load the data with the package**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities339-pop/](https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities339-pop/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
!pip install -q geo4bolivia
import geo4bolivia as geo
import plotly.express as px

pop = geo.load("pop")

# Totals by department, 2001-2020 (population sums up the hierarchy)
dep = geo.aggregate(pop, to="department", dataset="pop")
names = geo.units("department").set_index("geo_id")["name"]
dep["department"] = dep["geo_id"].map(names)
px.line(dep, x="year", y="pop", color="department",
labels={"pop": "Population"}, title="Population by department, 2001-2020")
```
