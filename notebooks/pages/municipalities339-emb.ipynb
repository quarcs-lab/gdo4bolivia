## Setup

Installs the packages every section uses and imports them once.

```python
!pip install -q geo4bolivia geolibre
import geo4bolivia as geo
import plotly.express as px
from sklearn.decomposition import PCA
import pandas as pd
import geopandas as gpd
from geolibre import Map
```

## Load the data with the package

```python
emb = geo.load("emb")              # A00-A63 by area, A00_pw-A63_pw by population
X = emb.filter(regex=r"^A\d{2}_pw$")             # the 64 population-weighted ones
emb["pc1"] = PCA(1).fit_transform(X)[:, 0]

# How the first component relates to multidimensional poverty (2012)
mpi = geo.load("sdg01", variables="sdg1_2_mpi", years=2012)
names = geo.units("map339")[["geo_id", "name"]]
df = emb[["geo_id", "pc1"]].merge(mpi, on="geo_id").merge(names, on="geo_id")
px.scatter(df, x="pc1", y="sdg1_2_mpi", hover_name="name", trendline="ols",
   labels={"pc1": "First principal component", "sdg1_2_mpi": "MPI, 2012"},
   title="Satellite embeddings and poverty")
```

## Load the data without the package (plain URLs)

```python
BASE = "https://raw.githubusercontent.com/quarcs-lab/gdo4bolivia/main/"
emb = pd.read_csv(BASE + "data/emb/map339/emb.csv")   # A00-A63, A00_pw-A63_pw

# Distribution of each of the 64 area-weighted dimensions across municipalities
px.box(emb.filter(regex=r"^A\d{2}$"),
       labels={"variable": "Dimension", "value": "Value"},
       title="Satellite embeddings, 2017")
```

## Load the data from the wide GeoPackage (one row per municipality)

```python
# No geo4bolivia needed: one row per municipality, a column per variable
# and year (<variable>_<year>), with boundaries
BASE = "https://raw.githubusercontent.com/quarcs-lab/gdo4bolivia/main/"
gdf = gpd.read_file(BASE + "data/emb/map339/emb_wide.gpkg",
                    layer="emb")   # or emb_wide.geojson (lighter)
px.choropleth_map(gdf, geojson=gdf.geometry, locations=gdf.index,
          color="A00_2017", hover_name="name",
          center={"lat": -16.3, "lon": -64}, zoom=4,
          labels={"A00_2017": "Embedding A00 (area-weighted)"},
          title="Embedding A00 (area-weighted), 2017")
```

## Load the data in GeoLibre (the wide GeoJSON)

```python
# GeoLibre in the notebook: the wide GeoJSON, one feature per municipality
# and a column per variable and year (<variable>_<year>); restyle any of
# them in the Style panel
BASE = "https://raw.githubusercontent.com/quarcs-lab/gdo4bolivia/main/"
m = Map(center=[-64.5, -16.5], zoom=4.5, height="600px")
m.add_choropleth(BASE + "data/emb/map339/emb_wide.geojson",
                 column="A00_2017", name="Embedding A00 (area-weighted), 2017",
                 strokeColor="#ffffff", strokeWidth=0.3)
m
```
