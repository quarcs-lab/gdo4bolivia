# expdpy: explore and model panel data

**Quickstart with the package**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/tools/expdpy-package/](https://quarcs-lab.github.io/gdo4bolivia/tools/expdpy-package/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
!pip install -q geo4bolivia expdpy
import geo4bolivia as geo
import expdpy as ex

df, df_def = geo.for_expdpy("sdg", variables="index_sdg",
                            groups=["department"])    # 343 units, 2 periods
df = ex.set_labels(df, df_def, set_panel=True)        # geo_id x year panel
display(ex.explore_xtsum_table(df, var=["index_sdg"]).gt)   # between vs within
ex.explore_trend_plot_by_group(df, group_var="department", var="index_sdg").fig
```
