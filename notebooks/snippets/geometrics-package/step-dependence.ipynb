# geometrics: regional growth, convergence and inequality

**Spatial dependence in each period**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/tools/geometrics-package/](https://quarcs-lab.github.io/gdo4bolivia/tools/geometrics-package/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
!pip install -q geo4bolivia geometrics
import geo4bolivia as geo
import geometrics as gm

gdf, df, df_dict = geo.for_geometrics("sdg", variables="index_sdg")
df = gm.set_labels(df, df_dict, set_panel=True)
w = gm.make_weights(gdf, method="queen")
for year in (2012, 2024):
    res = gm.explore_moran_plot(df, "index_sdg", gdf=gdf, w=w, period=year)
    print(year, res.interpret().split(".")[0])
gm.explore_lisa_cluster_map(df, "index_sdg", gdf=gdf, w=w, period=2012).fig
```
