# SDG 7: Affordable and clean energy

**Map the change, ~2012 to ~2024**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities343-sdg07/](https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities343-sdg07/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
!pip install -q geo4bolivia
import geo4bolivia as geo
import plotly.express as px

df = geo.load("sdg07", variables="index_sdg7")
wide = df.pivot(index="geo_id", columns="year", values="index_sdg7")
wide["change"] = wide[2024] - wide[2012]
gdf = geo.boundaries("map343").merge(wide["change"].reset_index(),
                                           on="geo_id")
px.choropleth_map(gdf, geojson=gdf.geometry, locations=gdf.index,
          color="change", hover_name="name",
          center={"lat": -16.3, "lon": -64}, zoom=4,
          labels={"change": "Change in the index (points)"},
          title="SDG 7: change in the index, ~2012 to ~2024")
```
