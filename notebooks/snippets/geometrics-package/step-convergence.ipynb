# geometrics: regional growth, convergence and inequality

**β-convergence, with and without space**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/tools/geometrics-package/](https://quarcs-lab.github.io/gdo4bolivia/tools/geometrics-package/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
!pip install -q geo4bolivia geometrics
import geo4bolivia as geo
import geometrics as gm

gdf, df, df_dict = geo.for_geometrics("sdg", variables="index_sdg")
df = gm.set_labels(df, df_dict, set_panel=True)
w = gm.make_weights(gdf, method="queen")
ols = gm.analyze_beta_convergence(df, "index_sdg")
sar = gm.analyze_beta_convergence(df, "index_sdg", model="sar", gdf=gdf, w=w)
print(ols.interpret())
print(sar.interpret())
sar.fig
```
