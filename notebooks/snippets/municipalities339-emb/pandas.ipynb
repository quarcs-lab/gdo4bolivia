# Satellite embeddings

**Load the data without the package (plain URLs)**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities339-emb/](https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities339-emb/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
import pandas as pd
import plotly.express as px

BASE = "https://raw.githubusercontent.com/quarcs-lab/gdo4bolivia/main/"
emb = pd.read_csv(BASE + "data/map339/embpw.csv")   # geo_id, year, asdf_id, A00-A63

# Distribution of each of the 64 dimensions across municipalities
px.box(emb.filter(regex=r"^A\d{2}$"),
       labels={"variable": "Dimension", "value": "Value"},
       title="Satellite embeddings, 2017")
```
