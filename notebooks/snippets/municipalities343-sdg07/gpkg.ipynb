# SDG 7: Affordable and clean energy

**Load the data from the wide GeoPackage (one row per municipality)**: the code of a box on [https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities343-sdg07/](https://quarcs-lab.github.io/gdo4bolivia/indicators/municipalities343-sdg07/) (GeoDevelopment Observatory of Bolivia), on its own. Run the cell (Runtime > Run all); it installs what it needs.

```python
import geopandas as gpd
import plotly.express as px

# No geo4bolivia needed: one row per municipality, a column per variable
# and year (<variable>_<year>), with boundaries
BASE = "https://raw.githubusercontent.com/quarcs-lab/gdo4bolivia/main/"
gdf = gpd.read_file(BASE + "data/sdg07/map343/sdg07_wide.gpkg",
                    layer="sdg07")   # or sdg07_wide.geojson (lighter)
px.choropleth_map(gdf, geojson=gdf.geometry, locations=gdf.index,
          color="index_sdg7_2024", hover_name="name",
          center={"lat": -16.3, "lon": -64}, zoom=4,
          labels={"index_sdg7_2024": "Affordable and clean energy index (0-100)"},
          title="Affordable and clean energy index (0-100), ~2024")
```
